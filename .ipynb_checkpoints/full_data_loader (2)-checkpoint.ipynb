# Make age and genotype groups based on pixelnmf_params (metadata file)

```python
def build_groups_from_excel(excel_path):
    
    # load metadata
    df = pd.read_excel(excel_path)

    # Ensure lowercase consistency (optional)
    df['genotype'] = df['genotype'].str.upper()

    # Dictionaries to store groups
    genotype_age_groups = defaultdict(list)
    age_only_groups = defaultdict(list)

    # Iterate and populate
    for _, row in df.iterrows():
        dataset_id = row['ds']
        age = row['age']
        genotype = row['genotype']

        # group by genotype + age (e.g., GABA_p2)
        genotype_age_groups[f"{genotype}_{age}"].append(dataset_id)

        # group by age (emx-gad) (e.g., all_p2)
        age_only_groups[f"all_{age}"].append(dataset_id)

    return dict(genotype_age_groups), dict(age_only_groups)
```

# Load MouseCraft

Load params xlsx as DataFrame.

Filter datasets where mousecraft == 1.

Parse dataset name to reconstruct path:

Extract first part sz089 (mouse ID).

Extract rest (date) and append path segments accordingly.

For each such dataset, open the validation_MF_final.xls.

Count twitches and active motions by:

Filtering status column to only include "accepted", "manually added", or "edited".

Summing counts and durations from onset–offset frames.

Convert frames to seconds using framerate and duration from your metadata.

```python
def load_mousecraft_validations_all(params_path, data_root, mode="cells"):
    """
    Scan all datasets under data_root/mode and load Mousecraft validation files if they exist.

    Returns a DataFrame with the results.
    """
    params_df = pd.read_excel(params_path)  # to read duration 
    
    # Handle different folder structures for cells vs pixels
    if mode == "cells":
        data_root = Path(data_root) / "cells"
    elif mode == "pixels":
        data_root = Path(data_root) / "pixels"
    else:
        data_root = Path(data_root) / mode

    results = {}
    
    # Get Mousecraft-enabled datasets from params_df
    if 'mousecraft' in params_df.columns:
        mousecraft_datasets = set(params_df[params_df['mousecraft'] == 1]['ds'])
        print(f"📊 Mousecraft-enabled datasets: {len(mousecraft_datasets)}")
    else:
        mousecraft_datasets = None
        print("⚠️ No 'mousecraft' column in params_df - processing all datasets")

    # Iterate over all mouse folders
    for mouse_folder in data_root.iterdir():
        if not mouse_folder.is_dir():
            continue

        # Iterate over all session folders
        for session_folder in mouse_folder.iterdir():
            if not session_folder.is_dir():
                continue

            # Build dataset name
            ds_name = f"{mouse_folder.name}_{session_folder.name}"
            
            # Skip if this dataset is not marked for Mousecraft processing
            if mousecraft_datasets and ds_name not in mousecraft_datasets:
                continue

            # Mousecraft validation path - handle different possible locations
            possible_validation_folders = [
                session_folder / "mousecraft_output",  # Standard location
                session_folder,  # Sometimes files are directly in session folder
                session_folder.parent / "mousecraft_output" / session_folder.name,  # Alternative structure
            ]
            
            valid_files = ["validation_HF_final.xlsx", "mousecraft_validated_labels_HF_final.xlsx"]
            
            validation_path = None
            for valid_folder in possible_validation_folders:
                for valid_file in valid_files:
                    candidate_path = valid_folder / valid_file
                    if candidate_path.exists():
                        validation_path = candidate_path
                        print(f'Found {valid_file} in {valid_folder}')  # MOVED BEFORE break
                        break  # Now this break happens after printing
                if validation_path:
                    break
                    
            if validation_path is None:
                print(f"⚠️ Missing Mousecraft file for {ds_name} at {session_folder}. Skipping.")
                continue

            print(f"✅ Found Mousecraft file for {ds_name}: {validation_path}")

            try:
                # Load Mousecraft validation
                val_df = pd.read_excel(validation_path)

                # Filter only accepted / edited / manually added
                valid_events = val_df[val_df['status'].isin(['accepted', 'edited', 'manually added'])]
                TP_events = val_df[val_df['status'].isin(['accepted', 'edited'])]
                FP_events = val_df[val_df['status'].isin(['rejected'])]
                TP_events_percentage =  len(TP_events)/len(TP_events) + len(FP_events)
                FP_events_percentage =  len(FP_events)/len(TP_events) + len(FP_events)
                performance_idx = (len(TP_events) - len(FP_events)) / (len(TP_events) + len(FP_events)) if (len(TP_events) + len(FP_events)) > 0 else np.nan
                
                # Twitches vs Active motions
                twitches = valid_events[valid_events['event_type'] == 'twitch']
                active = valid_events[valid_events['event_type'] == 'active']

                # ✅ Read total_time_sec (duration) from params_df
                total_time_sec = np.nan
                framerate = np.nan 

                if 'ds' in params_df.columns:
                    match = params_df.loc[params_df['ds'] == ds_name]
                    if not match.empty:
                        total_time_sec = match['duration'].iloc[0] if 'duration' in match.columns else np.nan
                        framerate = match['framerate'].iloc[0] if 'framerate' in match.columns else np.nan
                elif {'mouse', 'session'}.issubset(params_df.columns):
                    match = params_df.loc[
                        (params_df['mouse'] == mouse_folder.name) &
                        (params_df['session'] == session_folder.name)
                    ]
                    if not match.empty:
                        total_time_sec = match['duration'].iloc[0] if 'duration' in match.columns else np.nan
                        framerate = match['framerate'].iloc[0] if 'framerate' in match.columns else np.nan

                # Handle case where framerate might be missing
                if np.isnan(framerate):
                    print(f"⚠️ No framerate found for {ds_name}, using default 30 Hz")
                    framerate = 30.0

                
                twitch_count = len(twitches)
                active_motion_count = len(active)
                
                twitch_onsets = twitches['onset']
                twitch_offsets = twitches['offset']
                twitch_durations = ((twitch_offsets - twitch_onsets)).tolist()
                mean_twitch_duration = np.mean(twitch_durations)/framerate if twitch_count > 0 else 0
                
                inter_twitch_intrvl = np.diff(twitch_onsets).tolist() if twitch_count > 1 else []
                mean_inter_twitch_intrvl = np.mean(inter_twitch_intrvl) if len(inter_twitch_intrvl) > 0 else 0

                active_motion_onsets = active['onset']
                active_motion_offsets = active['offset']
                
                active_motion_durations = ((active_motion_offsets - active_motion_onsets)).tolist() if active_motion_count > 0 else []
                mean_active_motion_duration = np.mean(active_motion_durations)/framerate if active_motion_count > 0 else 0 #in sec
                active_motion_frequency = active_motion_count / total_time_sec if total_time_sec and total_time_sec > 0 else np.nan # freqency per duration

                
                inter_active_motion_intrvl = np.diff(active_motion_onsets).tolist() if active_motion_count > 1 else []
                mean_inter_active_motion_intrvl = np.mean(inter_active_motion_intrvl) if len(inter_active_motion_intrvl) > 0 else 0

                # Compute derived quantities
                active_motion_time_sec = np.sum(active_motion_durations)/framerate if active_motion_count > 0 else np.nan # sec 
                rest_time_sec = total_time_sec - active_motion_time_sec if not np.isnan(total_time_sec) else np.nan
                active_time_percentage = (active_motion_time_sec/total_time_sec) * 100
                rest_time_percentage = (rest_time_sec/total_time_sec) * 100
                twitch_frequency = twitch_count / rest_time_sec if rest_time_sec and rest_time_sec > 0 else np.nan


                # Store results
                results[ds_name] = {
                    "active_motion_count": active_motion_count,
                    "active_motion_frequency": active_motion_frequency,
                    "active_motion_durations": active_motion_durations,
                    "mean_active_motion_duration": mean_active_motion_duration,
                    "active_motion_intervals": inter_active_motion_intrvl,
                    "mean_active_motion_interval": mean_inter_active_motion_intrvl,
                    "active_mot_onsets": active_motion_onsets.tolist(),
                    "active_mot_offsets": active_motion_offsets.tolist(),

                    "twitch_count": twitch_count,
                    "twitch_durations": twitch_durations,
                    "mean_twitch_duration": mean_twitch_duration,
                    "twitch_intervals": inter_twitch_intrvl,
                    "mean_twitch_interval": mean_inter_twitch_intrvl,
                    "twitch_onsets": twitch_onsets.tolist(),
                    "twitch_offsets": twitch_offsets.tolist(),

                    "active_motion_time_sec": active_motion_time_sec,
                    "rest_time_sec": rest_time_sec,
                    "total_time_sec": total_time_sec,
                    'active_time_percentage': active_time_percentage, 
                    'rest_time_percentage':rest_time_percentage,
                    "twitch_frequency": twitch_frequency,
                    "framerate": framerate, 
                    "TP_events_percentage": TP_events_percentage,
                    "FP_events_percentage": FP_events_percentage,
                    "performance_idx": performance_idx
                }

            except Exception as e:
                print(f"❌ Error processing {ds_name}: {e}")
                continue

    if not results:
        print("❌ No Mousecraft validation data found!")
        return pd.DataFrame()

    result_df = (
        pd.DataFrame.from_dict(results, orient='index')
        .reset_index()
        .rename(columns={'index': 'ds'})
    )
    
    print(f"🎯 Successfully processed {len(result_df)} datasets")
    return result_df
```

# Load activity - behavior analysis 

```python
# load beh_dict 

def load_behavior_dicts_all(data_root, mode="cells"):
    """
    Traverse data_root/mode/mouse/session/behavior_outputs/beh_dict.npy
    and load all behavior dictionaries into a DataFrame.
    
    Parameters
    ----------
    data_root : str
        Path to the root of the data_proc folder
    mode : str
        Subfolder to search (default 'cells')
    
    Returns
    -------
    pd.DataFrame
        DataFrame of all beh_dict entries, merged into tabular format
    """
    
    beh_dicts = []
    base_path = os.path.join(data_root, mode)
    
    for mouse in os.listdir(base_path):
        mouse_path = os.path.join(base_path, mouse)
        if not os.path.isdir(mouse_path):
            continue
        
        for session in os.listdir(mouse_path):
            beh_path = os.path.join(mouse_path, session, "behavior_outputs", "beh_dict.npy")
            if os.path.exists(beh_path):
                try:
                    beh_dict = np.load(beh_path, allow_pickle=True).item()
                    # add ds key if not already there
                    if "ds" not in beh_dict:
                        beh_dict["ds"] = f"{mouse}_{session}"
                    beh_dicts.append(beh_dict)
                except Exception as e:
                    print(f"⚠️ Could not load {beh_path}: {e}")
    
    if not beh_dicts:
        print("⚠️ No beh_dict.npy files found!")
        return pd.DataFrame()
    
    return pd.DataFrame(beh_dicts)
```

```python
# Your root path to data_proc
base_path_cells = os.path.join(data_root, "cells")
base_path_pixels = os.path.join(data_root, "pixels")
base_path_dual = os.path.join(data_root, 'cells_dual_img')


all_beh_dicts = []
found_ds = []
missing_ds = []

# Process cells, pixels, and dual
for base_path, roi_type, merged_df in [
    (base_path_cells, "cells", merged_df_cells),
    (base_path_pixels, "pixels", merged_df_pixels),
    (base_path_dual, "dual", merged_df_dual)
]:
    datasets = merged_df["ds"].unique()

    for ds in datasets:
        parts = ds.split("_", 1)
        mouse_id, session_id = parts[0], parts[1]

        # Exact path first
        beh_path = os.path.join(base_path, mouse_id, session_id, "behavior_outputs", "beh_dict.npy")

        if not os.path.exists(beh_path):
            # fallback: replace '-' with '_'
            session_id_fixed = session_id.replace("-", "_")
            beh_path = os.path.join(base_path, mouse_id, session_id_fixed, "behavior_outputs", "beh_dict.npy")

        if os.path.exists(beh_path):
            beh_dict = np.load(beh_path, allow_pickle=True).item()
            beh_dict["ds"] = ds
            beh_dict["ROI_type"] = roi_type
            all_beh_dicts.append(beh_dict)
            found_ds.append(ds)
            print(f"✅ Found behavior data for {ds} ({roi_type})")
        else:
            missing_ds.append(ds)
            print(f"❌ Missing behavior data for {ds} ({roi_type})")

# Convert loaded behavior dicts to DataFrame
beh_df = pd.DataFrame(all_beh_dicts)

# Combine all ROI types into a single full_df
full_df = pd.concat([
    merged_df_cells.assign(ROI_type="cells"),
    merged_df_pixels.assign(ROI_type="pixels"),
    merged_df_dual.assign(ROI_type="dual")
], axis=0, ignore_index=True)

# Merge behavior data with full_df
full_df = pd.merge(full_df, beh_df, on=["ds", "ROI_type"], how="left")

# --- Summary ---
print(f"\n📊 Final Summary:")
print(f"✅ Found behavior_outputs for {len(found_ds)} datasets")
print(f"❌ Missing behavior_outputs for {len(missing_ds)} datasets")
print(f"📈 Total in full_df: {len(full_df)} rows")
print(f"   - Cells: {len(full_df[full_df['ROI_type'] == 'cells'])}")
print(f"   - Pixels: {len(full_df[full_df['ROI_type'] == 'pixels'])}")
print(f"   - Dual: {len(full_df[full_df['ROI_type'] == 'dual'])}")

# --- Optional: Check which datasets have behavior data ---
if len(beh_df.columns) > 2:  # More than just 'ds' and 'ROI_type'
    behavior_col = [col for col in beh_df.columns if col not in ["ds", "ROI_type"]][0]
    has_behavior = full_df[full_df[behavior_col].notna()]
    print(f"🔍 Rows with behavior data: {len(has_behavior)}")
else:
    print("🔍 No behavior columns found to check")
```

# Load SLEAP output 
For ds under cells or pixels and add in with full_df 
- load SLEAP, fit ellipse on points & extract eye_motion_x and eye_motion_y
- add in full_df

```python
import patchnmf.sleap_load_and_process as lp
from tqdm import tqdm
import h5py

def load_sleap_output(params_path, ds=None, data_root=None, file_format="auto"):
    """
    Load all eye points (ignore paw/snout/score) for a dataset.
    Now searches in both 'cells' and 'pixels' directories.
    Returns: eye_motion_x, eye_motion_y as 1D arrays or (None, None) if failed.
    """
    params_df = pd.read_excel(params_path)
    params_df['ds'] = params_df['ds'].astype(str).str.strip()

    ds_norm = ds.replace("-", "_").replace("/", "_").strip()
    row = params_df[params_df['ds'].str.replace("-", "_").str.replace("/", "_") == ds_norm]

    if row.empty or row['SLEAP'].values[0] != 1:
        return None, None

    row = row.iloc[0]
    parts = ds_norm.split("_")
    mouse_id = parts[0]
    session = "_".join(parts[1:])
    
    # Try both cells and pixels directories
    for mode in ["cells", "pixels", "cells_dual_img"]: # the 3 directories of all possible data 
        sleap_dir = Path(data_root) / mode / mouse_id / session / 'sleap_output'
        
        if not sleap_dir.exists():
            continue
            
        try:
            if file_format == "auto" or file_format == "csv":
                # Try CSV first
                csv_path = sleap_dir / "sleap_output.csv"
                if csv_path.exists():
                    df = pd.read_csv(csv_path).interpolate(method='linear')
                    
                    # Keep only eye columns ending with .x or .y, ignore .score
                    eye_columns = [c for c in df.columns if c.startswith("eye") and (c.endswith(".x") or c.endswith(".y"))]
                    if not eye_columns:
                        print(f"⚠️ No eye points found for {ds}")
                        continue
                        
                    coords_dict = lp.get_coordinates_dict(df, eye_columns)
                    
                elif file_format == "auto":
                    # Try H5 if CSV not found and auto mode
                    h5_files = list(sleap_dir.glob("*.h5"))
                    if h5_files:
                        with h5py.File(h5_files[0], "r") as f:
                            node_names = f['node_names'][:].astype(str)
                            tracks_matrix = f['tracks'][:]
                            coords_dict = {}
                            for i, name in enumerate(node_names):
                                if name.startswith("eye"):
                                    coords_dict[f"{name}.x"] = tracks_matrix[0, 0, i, :].squeeze()
                                    coords_dict[f"{name}.y"] = tracks_matrix[0, 1, i, :].squeeze()
                    else:
                        continue
                        
            elif file_format == "h5":
                h5_files = list(sleap_dir.glob("*.h5"))
                if not h5_files:
                    continue
                    
                with h5py.File(h5_files[0], "r") as f:
                    node_names = f['node_names'][:].astype(str)
                    tracks_matrix = f['tracks'][:]
                    coords_dict = {}
                    for i, name in enumerate(node_names):
                        if name.startswith("eye"):
                            coords_dict[f"{name}.x"] = tracks_matrix[0, 0, i, :].squeeze()
                            coords_dict[f"{name}.y"] = tracks_matrix[0, 1, i, :].squeeze()
            else:
                raise ValueError("Unsupported file format")

            # Only keep unique eyes
            unique_eyes = sorted({c.split(".")[0] for c in coords_dict.keys()})
            reformatted_dict = lp.get_reformatted_coordinates_dict(coords_dict, unique_eyes)

            # Interpolate NaNs
            for k, v in reformatted_dict.items():
                nans = np.isnan(v)
                if nans.any():
                    v[nans] = np.interp(np.flatnonzero(nans), np.flatnonzero(~nans), v[~nans])
                    reformatted_dict[k] = v

            # Fit ellipse using all available eyes
            ellipse_params, ellipse_centers = lp.get_fitted_ellipse_parameters(reformatted_dict, unique_eyes)
            if ellipse_centers.shape[0] == 0:
                continue

            eye_motion_x = ellipse_centers[:, 0]
            eye_motion_y = ellipse_centers[:, 1]
            
            print(f"✅ Loaded SLEAP data for {ds} from {mode} directory")
            return eye_motion_x, eye_motion_y

        except Exception as e:
            print(f"⚠️ Error processing SLEAP for {ds} in {mode}: {e}")
            continue

    print(f"❌ No SLEAP data found for {ds} in any directory")
    return None, None

def load_sleap(full_df, params_path, data_root, file_format="auto"):
    """
    Load SLEAP data for all datasets with SLEAP flag = 1
    """
    # Initialize columns
    for col in ["eye_motion_x", "eye_motion_y", "eye_displacement"]:
        full_df[col] = None
        full_df[col] = full_df[col].astype(object)

    for idx, row in tqdm(full_df.iterrows(), total=len(full_df), desc="Loading SLEAP outputs"):
        ds = row["ds"]
        sleap_flag = row.get("SLEAP", 0)

        if sleap_flag != 1:
            continue

        eye_motion_x, eye_motion_y = load_sleap_output(
            params_path=params_path,
            ds=ds,
            data_root=data_root,
            file_format=file_format
        )

        if eye_motion_x is None or eye_motion_y is None or len(eye_motion_x) != len(eye_motion_y):
            full_df.loc[idx, ["eye_motion_x", "eye_motion_y", "eye_displacement"]] = [None, None, None]
            continue

        # Calculate displacement
        dx = np.diff(eye_motion_x)
        dy = np.diff(eye_motion_y)
        eye_displacement = np.sqrt(dx**2 + dy**2)
        
        # Pad displacement to match original length (first value is 0)
        eye_displacement = np.concatenate(([0], eye_displacement))

        full_df.at[idx, "eye_motion_x"] = eye_motion_x.tolist()
        full_df.at[idx, "eye_motion_y"] = eye_motion_y.tolist()
        full_df.at[idx, "eye_displacement"] = eye_displacement.tolist()

    n_loaded = full_df["eye_motion_x"].apply(lambda x: x is not None and isinstance(x, list)).sum()
    print(f"✅ SLEAP data successfully loaded for {n_loaded} datasets")

    return full_df
```

```python
def load_sleap_and_motion_energy(full_df, params_path, data_root, file_format="auto"):
    """
    Load SLEAP eye data for all datasets in full_df.
    If Behaviour == 1, also load motion energy from camera_processed/motion_energy.npy.
    Adds columns: eye_motion_x, eye_motion_y, eye_displacement, motion_energy_original
    """
    from pathlib import Path
    import numpy as np
    from tqdm import tqdm

    # Initialize columns
    for col in ["eye_motion_x", "eye_motion_y", "eye_displacement", "motion_energy_original"]:
        full_df[col] = None
        full_df[col] = full_df[col].astype(object)

    for idx, row in tqdm(full_df.iterrows(), total=len(full_df), desc="Loading SLEAP and motion energy"):
        ds = row["ds"]
        sleap_flag = row.get("SLEAP", 0)
        behavior_flag = row.get("Behaviour", 0)

        # Initialize defaults
        eye_motion_x = eye_motion_y = eye_displacement = None
        motion_energy = None

        # Load SLEAP eye motion if available
        if sleap_flag == 1:
            eye_motion_x, eye_motion_y = load_sleap_output(
                params_path=params_path,
                ds=ds,
                data_root=data_root,
                file_format=file_format
            )
            if eye_motion_x is not None and eye_motion_y is not None and len(eye_motion_x) == len(eye_motion_y):
                dx = np.diff(eye_motion_x)
                dy = np.diff(eye_motion_y)
                eye_displacement = np.sqrt(dx**2 + dy**2)
                eye_displacement = np.concatenate(([0], eye_displacement))  # pad first value

        # Load motion energy if behaviour == 1
        if behavior_flag == 1:
            try:
                mouse_id, session = ds.split("_", 1)
                # Try both cells and pixels directories
                found = False
                for mode in ["cells", "pixels", "cells_dual_img"]:
                    motion_path = Path(data_root) / mode / mouse_id / session / "camera_processed" / "motion_energy.npy"
                    if motion_path.exists():
                        motion_energy = np.load(motion_path)
                        found = True
                        break
                if not found:
                    print(f"⚠️ motion_energy.npy not found for {ds} in cells or pixels folder")
            except Exception as e:
                print(f"⚠️ Error loading motion energy for {ds}: {e}")
                motion_energy = None

        # Store in dataframe
        full_df.at[idx, "eye_motion_x"] = eye_motion_x.tolist() if eye_motion_x is not None else None
        full_df.at[idx, "eye_motion_y"] = eye_motion_y.tolist() if eye_motion_y is not None else None
        full_df.at[idx, "eye_displacement"] = eye_displacement.tolist() if eye_displacement is not None else None
        full_df.at[idx, "motion_energy_original"] = motion_energy.tolist() if motion_energy is not None else None

    n_loaded_sleap = full_df["eye_motion_x"].apply(lambda x: x is not None and isinstance(x, list)).sum()
    n_loaded_me = full_df["motion_energy_original"].apply(lambda x: x is not None).sum()
    print(f"✅ SLEAP loaded for {n_loaded_sleap} datasets")
    print(f"✅ Motion energy loaded for {n_loaded_me} datasets")

    return full_df
```

```python
# add also motion aplitude computation 

def compute_motion_amplitudes(full_df, motion_col='motion_energy_norm',  #becs jures mot ens are in diff range (some ds)
                              active_onsets_col='active_mot_onsets', 
                              active_offsets_col='active_mot_offsets',
                              twitch_onsets_col='twitch_onsets_x',
                              twitch_offsets_col='twitch_offsets_x'):
    """
    Compute motion amplitude metrics (mean, peak, per state) from motion energy traces.
    """

    amp_results = []

    for _, row in full_df.iterrows():
        ds_name = row.get('ds', 'unknown')
        motion_energy = np.array(row[motion_col]) if isinstance(row[motion_col], (list, np.ndarray)) else None
        
        if motion_energy is None or len(motion_energy) == 0:
            amp_results.append({
                'ds': ds_name,
                'motion_amplitude_mean': np.nan,
                'motion_amplitude_peak': np.nan,
                'motion_amplitude_active': np.nan,
                'motion_amplitude_twitch': np.nan
            })
            continue

        # Base amplitudes
        motion_amp_mean = np.mean(motion_energy)
        motion_amp_peak = np.max(motion_energy)

        # Compute amplitude during active and twitch events
        def get_indices(onsets, offsets):
            if not isinstance(onsets, (list, np.ndarray)) or not isinstance(offsets, (list, np.ndarray)):
                return []
            idx = []
            for on, off in zip(onsets, offsets):
                on, off = int(on), int(off)
                idx.extend(range(on, min(off, len(motion_energy))))
            return np.array(idx, dtype=int)

        active_idx = get_indices(row[active_onsets_col], row[active_offsets_col])
        twitch_idx = get_indices(row[twitch_onsets_col], row[twitch_offsets_col])
        amplitudes_active = motion_energy[active_idx]
        amplitudes_twitch = motion_energy[twitch_idx] 
        motion_amp_active = np.mean(amplitudes_active) if len(active_idx) > 0 else np.nan
        motion_amp_twitch = np.mean(amplitudes_twitch) if len(twitch_idx) > 0 else np.nan
        
        # save results into lists 
        amp_results.append({
            'ds': ds_name,
            'motion_amplitude_mean': motion_amp_mean,
            'motion_amplitude_peak': motion_amp_peak,
            'amplitudes_active': amplitudes_active, 
            'amplitudes_twitch': amplitudes_twitch, 
            'motion_amplitude_active': motion_amp_active,
            'motion_amplitude_twitch': motion_amp_twitch
        })

    # add those columns into the full_df dataframe & drop any same named columns before
    amp_df = pd.DataFrame(amp_results)
    for col in ['motion_amplitude_mean', 'motion_amplitude_peak', 'motion_amplitude_active', 'motion_amplitude_twitch']:
        if col in full_df.columns:
            full_df = full_df.drop(columns=[col])
    
    full_df = full_df.merge(amp_df, on='ds', how='left')
    return full_df
```

```python
def normalize_motion_energy_per_dataset(full_df, col="motion_energy_original", new_col="motion_energy_norm"):
    """
    Normalize motion energy per dataset individually to 0-1.
    Skips None or empty traces.
    """
    full_df[new_col] = None
    
    for i, arr in enumerate(full_df[col]):
        if isinstance(arr, list) and len(arr) > 0:
            arr = np.array(arr, dtype=float)
            min_val = np.nanmin(arr)
            max_val = np.nanmax(arr)
            
            if max_val > min_val:
                arr_norm = (arr - min_val) / (max_val - min_val)
            else:
                arr_norm = arr  # constant signal
            
            full_df.at[i, new_col] = arr_norm.tolist()
    
    print(f"✅ Per-dataset normalized motion energies saved to column '{new_col}'")
    return full_df
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import os

def plot_metric_by_age(
    df,
    metric_col,
    age_col='age',
    ds_col='ds',
    kind='violin',        # 'violin', 'box', or 'line'
    show_points=True,
    cmap='Blues',
    figsize=(7,4),
    dpi=600,
    save_path=None,
    title=None,
    y_label=None,
    labels=None,
    ymin=None,
    ymax=None,
    poly_order=2  # polynomial order for line fit
):
    """
    Plot a metric by age:
    - 'violin', 'box', or 'line' style using Seaborn for smooth visuals
    - Shows per-mouse points and mean per age
    - Box/violin vertical lines represent ±1 SD
    - Removes grey Seaborn grid
    """
    df = df.copy()
    
    # Extract mouse_id
    df['mouse_id'] = df[ds_col].astype(str).str.split('_').str[0]

    # Normalize age labels (P7 = p7)
    def extract_age_num(label):
        match = re.search(r'(\d+)', str(label))
        return int(match.group(1)) if match else -1

    df['age_num'] = df[age_col].apply(extract_age_num)
    df['age_norm'] = df[age_col].str.upper()
    df = df[df['age_num'] >= 0]
    df = df.sort_values('age_num')

    # Unique ages and colormap
    unique_ages = sorted(df['age_norm'].unique(), key=lambda x: extract_age_num(x))
    num_ages = len(unique_ages)
    cmap_func = cm.get_cmap(cmap)
    age_colors = [cmap_func(i / max(1, num_ages-1)) for i in range(num_ages)]

    # Compute per-mouse mean
    per_mouse_means = df.groupby(['mouse_id', 'age_norm'])[metric_col].mean().reset_index(name='metric_mean')

    # Prepare plot data per age
    plot_data = [per_mouse_means[per_mouse_means['age_norm']==age]['metric_mean'].values
                 for age in unique_ages]
    means = [np.nanmean(vals) if len(vals) > 0 else np.nan for vals in plot_data]
    sds = [np.nanstd(vals) if len(vals) > 0 else np.nan for vals in plot_data]

    # --- Plot ---
    plt.figure(figsize=figsize, dpi=dpi)
    positions = np.arange(len(unique_ages))

    # Remove Seaborn grid
    sns.set_style("white")
    plt.grid(False)

    plot_df = pd.DataFrame({
        'value': np.concatenate(plot_data),
        'age': np.concatenate([[age]*len(vals) for age, vals in zip(unique_ages, plot_data)])
    })

    if kind in ['violin', 'box']:
        if kind == 'violin':
            sns.violinplot(x='age', y='value', data=plot_df,
                           palette=age_colors, inner=None, linewidth=0.3, width=0.35)
        else:
            sns.boxplot(x='age', y='value', data=plot_df,
                        palette=age_colors, width=0.35, fliersize=0, linewidth=0.3)

        # Overlay per-mouse points
        if show_points:
            sns.stripplot(x='age', y='value', data=plot_df, color='grey',
                          size=3, alpha=0.5, jitter=True)

        # Overlay mean ± SD lines and mean point
        for i, (mean_val, sd_val) in enumerate(zip(means, sds)):
            if not np.isnan(mean_val):
                plt.vlines(i, mean_val - sd_val, mean_val + sd_val, color='black', lw=1)
                plt.scatter(i, mean_val, color='black', s=7, zorder=5)

    else:
        # Line plot: polynomial fit through mean points
        valid_idx = ~np.isnan(means)
        if sum(valid_idx) > poly_order:
            coeffs = np.polyfit(positions[valid_idx], np.array(means)[valid_idx], deg=poly_order)
            x_fit = np.linspace(min(positions), max(positions), 300)
            y_fit = np.polyval(coeffs, x_fit)
            plt.plot(x_fit, y_fit, color='grey', linewidth=3, alpha=0.4, linestyle='--')

        if show_points:
            for i, mean_val in enumerate(means):
                if not np.isnan(mean_val):
                    plt.scatter(i, mean_val, color=age_colors[i], s=80, zorder=5)

    # Axes styling
    plt.xticks(positions, unique_ages)
    plt.axhline(0, linestyle='--', color='grey', lw=3, alpha=0.3)
    plt.ylabel(y_label or metric_col.replace('_',' ').capitalize())
    if ymin is not None or ymax is not None:
        plt.ylim(ymin, ymax)
    sns.despine()
    plt.tight_layout()
    if title:
        plt.title(title)

    # Save
    if save_path:
        os.makedirs(save_path, exist_ok=True)
        fname = (title or metric_col).replace(' ', '_') + '.png'
        plt.savefig(os.path.join(save_path, fname), dpi=dpi, bbox_inches='tight')
        print(f"✅ Saved figure to {os.path.join(save_path, fname)}")

    plt.show()


# plot two pairs of variables one next to each other per age tick (eg time % active - % rest)

def plot_metrics_by_age_side_by_side(
    df,
    metric_cols,  # list of 1 or 2 metrics
    age_col='age',
    ds_col='ds',
    figsize=(5,4),
    y_label=None,
    ymax=None,
    save_path=None,
    title=None,
    keep_p_labels=True,
    colors=('orange', 'blue'),
    kind='violin',  # 'violin' or 'box'
    show_points=True  # overlay per-mouse points
):
    """
    Plot the distribution of one or two metrics per age as violin or boxplots,
    aggregated per mouse, side-by-side per age, with optional per-mouse points.
    """

    if not isinstance(metric_cols, (list, tuple)):
        metric_cols = [metric_cols]

    df = df.copy()

    # --- Extract mouse ID ---
    df['mouse_id'] = df[ds_col].astype(str).str.split('_').str[0]

    # --- Clean and normalize age labels (P7 = p7 = 7) ---
    df['_age_str'] = df[age_col].astype(str).str.upper()
    df['_age_num'] = df['_age_str'].str.extract(r'(\d+)').astype(float)
    df = df[df['_age_num'].notna()]
    df = df.sort_values('_age_num')

    # --- Aggregate mean per mouse per age ---
    mouse_age_means = (
        df.groupby(['mouse_id', '_age_num'])[[*metric_cols]]
        .mean()
        .reset_index()
        .sort_values('_age_num')
    )
    mouse_age_means['age_str'] = ['P' + str(int(x)) for x in mouse_age_means['_age_num']]

    ages_sorted = mouse_age_means['age_str'].unique()
    n_ages = len(ages_sorted)

    # --- Plot setup ---
    plt.figure(figsize=figsize, dpi=300)
    width = 0.35  # width of each metric

    for i, metric in enumerate(metric_cols):
        color = colors[i] if i < len(colors) else 'gray'
        data = mouse_age_means.copy()
        data['metric'] = data[metric]

        # Shift positions for side-by-side plotting
        x_positions = np.arange(n_ages) + (i - 0.5) * width

        if kind == 'violin':
            for j, age in enumerate(ages_sorted):
                age_data = data[data['age_str'] == age]['metric']
                parts = plt.violinplot(
                    age_data,
                    positions=[x_positions[j]],
                    widths=width,
                    showmeans=False,
                    showmedians=False,
                    showextrema=False
                )
                for pc in parts['bodies']:
                    pc.set_facecolor(color)
                    pc.set_alpha(0.6)
                    pc.set_edgecolor('black')
        elif kind == 'box':
            for j, age in enumerate(ages_sorted):
                age_data = data[data['age_str'] == age]['metric']
                plt.boxplot(
                    age_data,
                    positions=[x_positions[j]],
                    widths=width,
                    patch_artist=True,
                    boxprops=dict(facecolor=color, edgecolor=color),
                    whiskerprops=dict(color=color),
                    capprops=dict(color=color),
                    medianprops=dict(color='black'),
                    showfliers=False
                )
        else:
            raise ValueError("`kind` must be 'violin' or 'box'")

        # --- Overlay per-mouse points ---
        if show_points:
            for j, age in enumerate(ages_sorted):
                age_data = data[data['age_str'] == age]['metric']
                x_jittered = np.random.normal(x_positions[j], width*0.05, size=len(age_data))
                plt.scatter(x_jittered, age_data, color='black', s=10, alpha=0.5)

    # --- Style ---
    plt.xticks(np.arange(n_ages), ages_sorted, rotation=45, fontsize=10)
    plt.title(title or f"{' & '.join(metric_cols)} per Age")
    plt.xlabel('Age')
    plt.ylabel(y_label or 'Metric')

    if ymax:
        plt.ylim(0, ymax)
    sns.despine()
    plt.tight_layout()

    # --- Save or show ---
    if save_path:
        os.makedirs(os.path.dirname(save_path), exist_ok=True)
        fname = os.path.join(save_path, (title or "_".join(metric_cols)) + ".png")
        plt.savefig(fname, dpi=300, bbox_inches='tight')
        print(f"✅ Saved plot to {fname}")
    else:
        plt.show()

from scipy.stats import gaussian_kde


def plot_metrics_evolution(
    df,
    metric_cols,  # list of 1 or 2 metrics
    age_col='age',
    ds_col='ds',
    y_label=None,
    ymax=None,
    save_path=None,
    title=None,
    labels=None,
    colors=('Blues','Oranges'),  # can be a colormap name or tuple of fixed colors
    sigma=2  # smoothing for Gaussian filter
):
    """
    Plot evolution of 1 or 2 metrics per age:
    - Mean per mouse per age as scatter points
    - Gaussian-smoothed line (optional)
    - Color-coded either by fixed colors or colormaps
    """
    sns.set_theme(style="white")

    if not isinstance(metric_cols, (list, tuple)):
        metric_cols = [metric_cols]

    df = df.copy()
    df['mouse_id'] = df[ds_col].astype(str).str.split('_').str[0]

    # Convert age to numeric
    if df[age_col].dtype == object:
        df[age_col] = df[age_col].astype(str).str.extract(r'(\d+)')[0].astype(float)

    df = df[df[age_col].notna()]
    ages_sorted = np.sort(df[age_col].unique())
    positions = np.arange(len(ages_sorted))  # for jitter if needed

    plt.figure(figsize=(7, 4), dpi=300)

    if labels is None:
        labels = metric_cols

    for i, metric in enumerate(metric_cols):
        color = colors[i] if i < len(colors) else 'gray'
        label = labels[i] if i < len(labels) else metric

        # --- Compute per-mouse mean per age ---
        mouse_age_means = df.groupby(['mouse_id', age_col])[metric].mean().reset_index()

        # --- Compute mean ± SD across mice per age ---
        mean_per_age = mouse_age_means.groupby(age_col)[metric].mean()
        sd_per_age = mouse_age_means.groupby(age_col)[metric].std()

        # --- Scatter mean per age ---
        # If color is a colormap, assign one color per age
        if isinstance(color, str) and color in plt.colormaps():
            cmap = plt.get_cmap(color)
            age_colors = [cmap(j / max(len(ages_sorted)-1,1)) for j in range(len(ages_sorted))]
        else:
            age_colors = [color]*len(ages_sorted)

        for j, age in enumerate(ages_sorted):
            plt.scatter(age, mean_per_age.loc[age], color=age_colors[j], s=100, alpha=0.9, zorder=5)

        # --- Optional Gaussian-smoothed trend line ---
        x_fit = np.linspace(min(ages_sorted), max(ages_sorted), 300)
        y_fit = np.interp(x_fit, mean_per_age.index, mean_per_age.values)
        y_fit_smooth = gaussian_filter1d(y_fit, sigma=sigma)
        plt.plot(x_fit, y_fit_smooth, color='grey', linewidth=3, alpha=0.3, zorder=4)

    # --- Labels, title, legend ---
    plt.xlabel('Age (days)')
    plt.ylabel(y_label or 'Metric')
    plt.ylim(0,ymax)
    plt.title(title or 'Metric evolution with age')
    plt.legend(labels, frameon=False, bbox_to_anchor=(1.01,1), loc='upper left')
    sns.despine()
    plt.tight_layout()

    if save_path:
        os.makedirs(save_path, exist_ok=True)
        fname = f"{title.replace(' ','_')}.png"
        plt.savefig(os.path.join(save_path, fname), dpi=300, bbox_inches='tight')
        print(f"✅ Saved plot to {os.path.join(save_path, fname)}")
    else:
        plt.show()
```

```python
def batch_pearson(signal1, signal2, bin_mask):
    """
    Compute Pearson correlation per contiguous batch in bin_mask (1 or 0),
    return mean across batches and optionally all per-batch correlations.
    """
    batch_corrs = []
    in_batch = False
    start_idx = 0
    for i, val in enumerate(bin_mask):
        if val == 1 and not in_batch:
            # start of a batch
            in_batch = True
            start_idx = i
        elif val == 0 and in_batch:
            # end of a batch
            end_idx = i
            if end_idx - start_idx > 1:  # avoid single-point batches
                batch_corrs.append(np.corrcoef(signal1[start_idx:end_idx],
                                            signal2[start_idx:end_idx])[0,1])
            in_batch = False
    # catch last batch
    if in_batch:
        end_idx = len(bin_mask)
        if end_idx - start_idx > 1:
            batch_corrs.append(np.corrcoef(signal1[start_idx:end_idx],
                                        signal2[start_idx:end_idx])[0,1])
    if len(batch_corrs) > 0:

        return np.nanmean(batch_corrs), batch_corrs
    else:
        return np.nan, []
```

```python
from scipy import signal 
#Gaussian smoothing
smooth_sigma = 5 # for computing correlation 

results = []

# Filter sessions with SLEAP == 1 and motion energy also, ** careful needs local xls and drive xls to be up to date 
df_sleap = full_df[(full_df['SLEAP'] == 1) & (full_df['motion_energy_original'].notna())] # we need the sessions that contain awakeness 

# for those sessions, load eye variables (after ellipse fit), motion energy, framerate, sth else ?
for idx, row in df_sleap.iterrows():
    sess_name = row['ds']
    age = row.get('age', np.nan)

    
    # Extract as numpy arrays
    eye_x = np.array(row['eye_motion_x'])
    eye_y = np.array(row['eye_motion_y'])
    eye_disp = np.array(row['eye_displacement'])

    x_vel = pad(np.diff(eye_x))
    y_vel = pad(np.diff(eye_y))
    eye_velocity = np.abs(x_vel) + np.abs(y_vel) # linear sum of x and y velocities 
    
    mot_en_raw = row.get('motion_energy_original', None)
    if mot_en_raw is None:
        print(f"Skipping {sess_name} (motion_energy_original missing)")
        continue

    mot_en_raw = np.array(mot_en_raw)
    mot_en = pad(mot_en_raw)  # assuming pad() returns a np.array
    if mot_en is None or mot_en.size == 0:
        print(f"Skipping {sess_name} (motion_energy_original empty after padding)")
        continue
        
    active_onsets = np.array(row['active_mot_onsets'])
    active_offsets = np.array(row['active_mot_offsets'])
    bin_mot_en = np.zeros_like(mot_en)
    age = row.get('age', np.nan)
    mouse_id = row.get('ds', np.nan)


    for onset, offset in zip(active_onsets, active_offsets):
        bin_mot_en[onset:offset]= 1
        
    twitch_onsets = np.array(row['twitch_onsets_x']) # this _x is an issue from sleap loader - remove 
    twitch_offsets = np.array(row['twitch_offsets_x'])

    min_len = min(len(eye_disp), len(mot_en), len(eye_x), len(eye_y)) # cut to the shortest to align 
    eye_disp = eye_disp[:min_len]
    eye_x = eye_x[:min_len]
    eye_y = eye_y[:min_len]

    mot_en = mot_en[:min_len]
    bin_mot_en = bin_mot_en[:min_len]

    mean_activity = np.array(row.get('data_mean_orig', np.zeros_like(mot_en)))
    framerate = row.get('framerate_orig')
    if framerate is None or np.isnan(framerate):
        framerate = 15 # fallback, here for eye all are imaged at 15 Hz
    else:
        framerate = int(framerate)
    # if exists then:
    if mot_en is not None:
        mot_en_smooth = gaussian_filter1d(mot_en, smooth_sigma)
    else:
        mot_en_smooth = None

    if eye_disp is not None:
        eye_disp_smooth = gaussian_filter1d(eye_disp, smooth_sigma)
    else:
       eye_disp_smooth = None

    # --- Compute correlations ---
    if mot_en is not None and eye_disp is not None:
        pear_corr, pear_p = pearsonr(eye_disp_smooth, mot_en_smooth)
    # spear_corr, spear_p = spearmanr(eye_disp_smooth, mot_en_smooth)

    # --- Cross-correlation ---
    corr = signal.correlate(
        eye_disp_smooth - np.mean(eye_disp_smooth),
        mot_en_smooth - np.mean(mot_en_smooth),
        mode='full'
    )
    lags = np.arange(-len(mot_en)+1, len(mot_en))
    lags_sec = lags/framerate
    corr /= np.max(corr)  # normalize to [-1, 1]

    lag_window = int(framerate*5)  # e.g., ±3 seconds
    mask = (lags >= -lag_window+1) & (lags <= lag_window+1)

    # displacement in x or y 
    dx = np.diff(eye_x, prepend=eye_x[0])
    dy = np.diff(eye_y, prepend=eye_y[0])
    abs_dx = np.abs(dx)
    abs_dy = np.abs(dy)
    mean_dx = np.mean(abs_dx) # only positive ! 
    mean_dy = np.mean(abs_dy) # only positive (negative space does not exist for arrow)
    direction = "X" if mean_dx > mean_dy else "Y"

    # peak detection per state
    peaks_active, _ = find_peaks(eye_disp_smooth[bin_mot_en==1])
    peaks_rest, _ = find_peaks(eye_disp_smooth[bin_mot_en==0])
    freq_active = len(peaks_active) / (np.sum(bin_mot_en==1)/framerate/60) if np.sum(bin_mot_en==1)>0 else np.nan
    freq_rest = len(peaks_rest) / (np.sum(bin_mot_en==0)/framerate/60) if np.sum(bin_mot_en==0)>0 else np.nan
    
    # --- Plot ---
    fig, axes = plt.subplots(6, 1, figsize=(10, 10), sharex=False, dpi=300)

    axes[0].plot(eye_x, color='gray')
    axes[0].set_ylabel('eye x')
    sns.despine(ax=axes[0])

    axes[1].plot(eye_y, color='gray')
    axes[1].set_ylabel('eye y')
    sns.despine(ax=axes[1])

    axes[2].plot(eye_velocity)
    # axes[2].plot(eye_disp_smooth, color='blue', label='smoothed')
    axes[2].set_ylabel('eye velocity')
    axes[2].legend(frameon=False)
    sns.despine(ax=axes[2])

    axes[3].plot(mot_en)
    # axes[3].plot(mot_en_smooth, color='red', label='smoothed')
    axes[3].set_ylabel('mot en')
    axes[3].legend(frameon=False)
    sns.despine(ax=axes[3])

    axes[4].plot(mean_activity)
    axes[4].set_ylabel('mean acts')
    sns.despine(ax=axes[4])

    axes[5].plot(lags_sec[mask], corr[mask])
    axes[5].axvline(0, color='gray', linestyle='--')
    axes[5].set_xlabel('Lag (frames)')
    axes[5].set_ylabel('Cross-corr')
    sns.despine(ax=axes[5])

    plt.suptitle(
        f"Session: {sess_name}\nPearson r = {pear_corr:.2f}")
    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, 'eye_motion_mot_en.png'), dpi=300)
    plt.show()

    # plot eye motion per direction per mouse 
    plt.figure(figsize=(3,2), dpi=300)
    sc = plt.scatter(eye_x, eye_y, s=1, c=mot_en,cmap='Reds', alpha=0.7)
    plt.colorbar(sc, label='mot_en')
    plt.show()
    
    plt.figure(figsize=(3,2), dpi=300)
    sc = plt.scatter(dx, dy, s=1, c=mot_en,cmap='Reds', alpha=0.7)
    plt.colorbar(sc, label='mot_en')
    plt.show()
    
    # bin motion energy
    
    cmap = ListedColormap(['blue', 'orange'])

    plt.figure(figsize=(3,2), dpi=300)
    sc = plt.scatter(eye_x, eye_y, s=3, c=bin_mot_en, cmap=cmap, alpha=0.5)
    plt.colorbar(sc, label='bin_mot_en')
    plt.show()
    
    plt.figure(figsize=(3,2), dpi=300)
    sc = plt.scatter(dx, dy, s=3, c=bin_mot_en, cmap=cmap, alpha=0.5)
    plt.colorbar(sc, label='bin_mot_en')
    plt.show()

    results.append({
        'session': sess_name,
        'mouse': mouse_id,
        'age': age,
        'pearson_r': pear_corr,
        'pearson_p': pear_p
    })
    
    corr_df = pd.DataFrame(results)
```

```python
import matplotlib.patches as mpatches
from scipy import signal

smooth_sigma = 3  # Gaussian smoothing
window_sec_twitch = 1  # time window after twitch
window_sec_active = 1  # time window after twitch

results = []

# Filter sessions
df_sleap = full_df[(full_df['SLEAP'] == 1) & (full_df['motion_energy_original'].notna())]

for idx, row in tqdm(df_sleap.iterrows(), total=len(df_sleap)):
    sess_name = row['ds']
    age = row.get('age', np.nan)
    mouse_id = row.get('ds', sess_name)

    def safe_array(x):
        if x is None:
            return np.array([])
        x = np.array(x)
        if x.ndim == 0:  # scalar
            return np.array([x])
        return x
    
    eye_x = safe_array(row.get('eye_motion_x'))
    eye_y = safe_array(row.get('eye_motion_y'))
    eye_disp = safe_array(row.get('eye_displacement'))
    mot_en_raw = row.get('motion_energy_original', None)

    if eye_x is None or len(eye_x) == 0 or eye_y is None or len(eye_y) == 0:
        print(f"⚠️ Skipping row: missing or empty eye tracking data.")
        continue
    
    x_velocity = pad(np.diff(eye_x))
    y_velocity = pad(np.diff(eye_y))

    abs_dx = np.abs(x_velocity)
    abs_dy = np.abs(y_velocity)
    
    eye_velocity = abs_dx + abs_dy # linear sum of velocity (abs values, non-directional)
    eye_acceleration = pad(np.diff(eye_velocity))
    
    # Mean displacement overall
    mean_dx = np.mean(abs_dx) # for arrows, this is by default non negative (cannot go in neg. directions)
    mean_dy = np.mean(abs_dy)
        
    if mot_en_raw is None:
        print(f"Skipping {sess_name}: motion_energy_original missing")
        continue
    mot_en = safe_array(pad(np.array(mot_en_raw)))
    
    if mot_en_raw is None:
        print(f"Skipping {sess_name}: motion_energy_original missing")
        continue
        
    mot_en = pad(np.array(mot_en_raw))
    if mot_en is None or mot_en.size == 0 or eye_velocity.size == 0:
        print(f"Skipping {sess_name}: empty data")
        continue

    # Align lengths
    min_len = min(len(eye_velocity), len(mot_en), len(eye_x), len(eye_y), len(abs_dx), len(abs_dy))
    eye_disp = eye_disp[:min_len]
    eye_x = eye_x[:min_len]
    dx = abs_dx[:min_len]
    dy =abs_dy[:min_len]
    eye_y = eye_y[:min_len]
    mot_en = mot_en[:min_len]
    eye_velocity = eye_velocity[:min_len]
    
    # Binary motion energy
    active_onsets = np.array(row.get('active_mot_onsets', []))
    active_offsets = np.array(row.get('active_mot_offsets', []))
    
    bin_mot_en = np.zeros_like(mot_en)
    for onset, offset in zip(active_onsets, active_offsets):
        bin_mot_en[onset:offset] = 1

    twitch_onsets = np.array(row.get('twitch_onsets_x', []))  # fix !! 
    twitch_offsets = np.array(row.get('twitch_offsets_x', []))
    
    twitch_mot_en = np.zeros_like(mot_en)
    for onset, offset in zip(twitch_onsets, twitch_offsets):
        twitch_mot_en[onset:offset] = 1

    # Smooth signals
    mot_en_smooth = gaussian_filter1d(mot_en, smooth_sigma)
    eye_disp_smooth = gaussian_filter1d(eye_disp, smooth_sigma)
    eye_velocity_smooth = gaussian_filter1d(eye_velocity, smooth_sigma)

    from scipy.stats import zscore
    eye_velocity_smooth_z = zscore(eye_velocity_smooth)
    mot_en_smooth_z = zscore(mot_en_smooth)
    pear_corr = np.corrcoef(eye_velocity_smooth_z, mot_en_smooth_z)[0,1]
    
    # Pearson correlation
    # pear_corr = np.corrcoef(eye_velocity_smooth, mot_en_smooth)[0,1] #only upper traingle - diagonal

    # Cross-correlation
    cross_corr = correlate(eye_velocity_smooth_z - np.mean(eye_velocity_smooth_z),
                           mot_en_smooth_z - np.mean(mot_en_smooth_z),
                           mode='full') # normalise 
    
    lags = np.arange(-len(mot_en) + 1, len(mot_en))

    framerate = row.get('framerate_orig', 15)
    framerate = int(framerate) if not pd.isna(framerate) else 15
    
    lags_sec = lags / framerate
    cross_corr /= np.max(cross_corr)
    lag_window = int(framerate * 3) # 3s 
    mask = (lags >= -lag_window) & (lags <= lag_window)

        # --- Plot ---
    fig, axes = plt.subplots(6, 1, figsize=(10, 10), sharex=False, dpi=300)

    axes[0].plot(eye_x, color='gray')
    axes[0].set_ylabel('eye x')
    sns.despine(ax=axes[0])

    axes[1].plot(eye_y, color='gray')
    axes[1].set_ylabel('eye y')
    sns.despine(ax=axes[1])

    axes[2].plot(eye_velocity_smooth)
    axes[2].set_ylabel('eye velocity')
    sns.despine(ax=axes[2])

    axes[3].plot(mot_en_smooth)
    axes[3].set_ylabel('motion energy')
    sns.despine(ax=axes[3])

    axes[4].plot(eye_disp_smooth)
    axes[4].set_ylabel('eye displacement')
    sns.despine(ax=axes[4])

    axes[5].plot(lags_sec[mask], cross_corr[mask])
    axes[5].axvline(0, color='gray', linestyle='--')
    axes[5].set_xlabel('Lag (sec)')
    axes[5].set_ylabel('Cross-correlation')
    sns.despine(ax=axes[5])

    plt.suptitle(f"Session: {sess_name}\nPearson r = {pear_corr:.2f}")
    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, f'{idx}.png'))
    plt.show()

    # per state correlation
    pear_active_mean, pear_active_batches = batch_pearson( 
        eye_velocity_smooth, mot_en_smooth, bin_mot_en
    ) # only pear active mean inter corrs why ?
    
    pear_rest_mean, pear_rest_batches = batch_pearson(
        eye_velocity_smooth, mot_en_smooth, 1 - bin_mot_en
    )

    #     # --- Pearson correlation during twitches ---
    # pear_twitch_mean, pear_twitch_batches = batch_pearson(
    #     eye_velocity_smooth, mot_en_smooth, twitch_mot_en
    # )

    if np.sum(twitch_mot_en) > 1:
        pear_twitch_mean = np.corrcoef(
            eye_velocity_smooth[twitch_mot_en == 1],
            mot_en_smooth[twitch_mot_en == 1]
        )[0,1]
    else:
        pear_twitch_mean = np.nan

    print('Corr eye motion - mot_en:', pear_corr)
    print("Mean Pearson active:", pear_active_mean)
    print("Mean Pearson rest:", pear_rest_mean)
    print("Mean Pearson twitch:", pear_twitch_mean)

    # Peak detection per state
    peaks_active, _ = find_peaks(eye_velocity_smooth[bin_mot_en == 1], prominence= np.std(eye_velocity_smooth) * 1.5)
    peaks_rest, _ = find_peaks(eye_velocity_smooth[bin_mot_en == 0], prominence= np.std(eye_velocity_smooth) * 1.5) # sd 
    dur_active_min = np.sum(bin_mot_en == 1) / framerate / 60
    dur_rest_min = np.sum(bin_mot_en == 0) / framerate / 60
    freq_active = len(peaks_active) / dur_active_min if dur_active_min > 0 else np.nan
    freq_rest = len(peaks_rest) / dur_rest_min if dur_rest_min > 0 else np.nan


    # plot peak detection output 

    active_indices = np.where(bin_mot_en == 1)[0]  # original indices of active frames
    rest_indices = np.where(bin_mot_en == 0)[0]    # original indices of rest frames

    peaks_active_global = active_indices[peaks_active]
    peaks_rest_global = rest_indices[peaks_rest]

    # Plot Active vs Rest Peaks
    plt.figure(figsize=(7,3), dpi=300)
    fig, ax = plt.subplots()
    ax.plot(eye_velocity_smooth, label="Eye Velocity")
    ax.scatter(peaks_active_global, eye_velocity_smooth[peaks_active_global], color='red', label='Active Peaks')
    ax.scatter(peaks_rest_global, eye_velocity_smooth[peaks_rest_global], color='blue', label='Rest Peaks')
    ax.set_xlabel("Time (frames)")
    ax.set_ylabel("Eye Velocity")
    ax.set_title(f"Active vs Rest Peaks: {sess_name}")
    ax.legend()

    
    # twitch reliability to trigger eye 
    window_frames_twitch = int(window_sec_twitch * framerate)
    window_frames_active = int(window_sec_active * framerate)

    max_disp_list = []
    mean_disp_list = []
    moved_flag = []
    baseline_std = np.std(eye_velocity_smooth)
    
    for onset in twitch_onsets:
        start = onset
        end = min(onset + window_frames_twitch, len(eye_disp_smooth))
        segment = eye_velocity_smooth[start:end]
        if len(segment) > 0:
            max_disp = np.max(segment)
            mean_disp = np.mean(segment)
            moved = max_disp > 1 * baseline_std
            max_disp_list.append(max_disp)
            mean_disp_list.append(mean_disp)
            moved_flag.append(moved)
    
    fraction_moved = np.mean(moved_flag) if len(moved_flag) > 0 else np.nan

    # inversely how many eye mot peaks are synched with twitches 

        # --- Synchrony of eye movement peaks with twitches during rest ---
    sync_window_sec = 1  # window around twitch onset to count as 'synchronized'
    window_frames_sync = int(sync_window_sec * framerate)

    # Convert rest peaks (masked) to global indices
    rest_mask_indices = np.where(bin_mot_en == 0)[0]
    active_mask_indices = np.where(bin_mot_en ==1)[0]
    peaks_rest_global = rest_mask_indices[peaks_rest] if len(peaks_rest) > 0 else np.array([])
    peaks_active_global = active_mask_indices[peaks_active] if len(peaks_active) > 0 else np.array([])
    
    synched_count_followed = 0
    for pr in peaks_rest_global: # prospective twitches 
        if np.any(((twitch_onsets - pr) >0) & (twitch_onsets - pr <= window_frames_sync)):
            synched_count_followed += 1
    # prospective eye twitches followed within 1 sec by twitch onset
    fraction_eye_followed_twitch_onset = synched_count_followed / len(peaks_rest_global) if len(peaks_rest_global) > 0 else np.nan

    synched_count_followed_active = 0
    for pr in peaks_active_global: # prospective twitches 
        if np.any(((active_onsets - pr) >0) & (active_onsets - pr <= window_frames_sync)):
            synched_count_followed_active += 1
    # prospective eye twitches followed within 1 sec by twitch onset
    fraction_eye_followed_active_onset = synched_count_followed_active / len(peaks_active_global) if len(peaks_active_global) > 0 else np.nan

    synched_count_preceded = 0
    for pr in peaks_rest_global:
        if np.any(((twitch_onsets - pr) < 0) & (pr - twitch_onsets <= window_frames_sync)):
            synched_count_preceded += 1
    # prospective eye twitches followed within 1 sec by twitch onset
    fraction_eye_preceded_twitch_onset = synched_count_preceded / len(peaks_rest_global) if len(peaks_rest_global) > 0 else np.nan

    synched_count_preceded_active = 0
    for pr in peaks_active_global: # prospective twitches 
        if np.any(((active_onsets - pr) < 0) & (pr - active_onsets <= window_frames_sync)):
            synched_count_preceded_active += 1
    # prospective eye twitches followed within 1 sec by twitch onset
    fraction_eye_preceded_active_onset = synched_count_preceded_active / len(peaks_active_global) if len(peaks_active_global) > 0 else np.nan

    active_max_disp_list = []
    active_mean_disp_list = []
    moved_flag = []
    
    for onset in active_onsets:
        start = onset
        end = min(onset + window_frames_active, len(eye_disp_smooth))
        segment = eye_velocity_smooth[start:end]
        if len(segment) > 0:
            max_disp = np.max(segment)
            mean_disp = np.mean(segment)
            moved = max_disp > 1 * baseline_std # threshold moved or not moved 
            active_max_disp_list.append(max_disp)
            active_mean_disp_list.append(mean_disp)
            moved_flag.append(moved)
    
    active_fraction_moved = np.mean(moved_flag) if len(moved_flag) > 0 else np.nan
    
    # --- Mean dx/dy per state ---
    if np.sum(bin_mot_en==1) > 0:
        mean_dx_active = np.mean(abs_dx[bin_mot_en==1])
        mean_dy_active = np.mean(abs_dy[bin_mot_en==1])

    if np.sum(bin_mot_en==0) > 0:
        mean_dx_rest = np.mean(abs_dx[bin_mot_en==0])
        mean_dy_rest = np.mean(abs_dy[bin_mot_en==0])
    else:
        mean_dx_rest = np.nan
        mean_dy_rest = np.nan

    # --- Eye trajectory scatterplot with twitches highlighted ---
    fig2, ax2 = plt.subplots(figsize=(4, 4), dpi=300)

    colors = np.array(['blue', 'orange'])[bin_mot_en.astype(int)]
    ax2.scatter(x_velocity, y_velocity, s=2, alpha=0.3, c=colors, label='all time')

    # Highlight twitch frames in red
    twitch_mask = twitch_mot_en == 1 
    ax2.scatter(x_velocity[twitch_mask], y_velocity[twitch_mask], s=2, color='red', alpha=0.5, label='twitch')

    ax2.set_xlabel('Eye X velocity')
    ax2.set_ylabel('Eye Y velocity')
    ax2.set_title(f"Eye velocity with twitches — {sess_name}")
    ax2.legend(frameon=False)
    sns.despine(ax=ax2)

    plt.savefig(os.path.join(save_dir, f'{idx}_eye_velocity_states_scatter.png'))
    plt.tight_layout()
    plt.show()

        # plot eye motion per direction per mouse 
    plt.figure(figsize=(3,2), dpi=300)
    sc = plt.scatter(eye_x, eye_y, s=1, c=mot_en,cmap='Reds', alpha=0.7)
    plt.colorbar(sc, label='mot_en')
    plt.show()
    
    plt.figure(figsize=(3,2), dpi=300)
    sc = plt.scatter(dx, dy, s=1, c=mot_en,cmap='Reds', alpha=0.7)
    plt.colorbar(sc, label='mot_en')
    plt.show()
    
    # bin motion energy
    
    cmap = ListedColormap(['blue', 'orange'])

    plt.figure(figsize=(3,2), dpi=300)
    sc = plt.scatter(eye_x, eye_y, s=3, c=bin_mot_en, cmap=cmap, alpha=0.5)
    plt.colorbar(sc, label='bin_mot_en')
    plt.show()
    
    plt.figure(figsize=(3,2), dpi=300)
    sc = plt.scatter(dx, dy, s=3, c=bin_mot_en, cmap=cmap, alpha=0.5)
    plt.colorbar(sc, label='bin_mot_en')
    plt.show()


    eye_velocity_per_sec = eye_velocity_smooth * framerate # pixels/sec on smooth mot_en 
    mean_vel_twitch = np.mean(eye_velocity_per_sec[twitch_mot_en==1])
    mean_vel_rest = np.mean(eye_velocity_per_sec[(twitch_mot_en == 0) & (bin_mot_en == 0)])
    mean_vel_active = np.mean(eye_velocity_per_sec[(twitch_mot_en == 0) & (bin_mot_en == 1)])
    
    # Gather data into a tidy DataFrame use eye_velocity smooth or eve_velocity (same as in peak detection so smooth makes sense)
    velocities = pd.DataFrame({
        'velocity': np.concatenate([
            eye_velocity_per_sec[twitch_mot_en == 1],
            eye_velocity_per_sec[(twitch_mot_en == 0) & (bin_mot_en == 0)],
            eye_velocity_per_sec[(twitch_mot_en == 0) & (bin_mot_en == 1)]
        ]),
        'state': np.concatenate([
            np.repeat('twitch', np.sum(twitch_mot_en == 1)),
            np.repeat('rest', np.sum((twitch_mot_en == 0) & (bin_mot_en == 0))),
            np.repeat('active', np.sum((twitch_mot_en == 0) & (bin_mot_en == 1)))
        ])
    })
    
    plt.figure(figsize=(4,4), dpi=300)
    sns.boxplot(data=velocities, x='state', y='velocity', palette={'rest':'blue', 'active':'orange', 'twitch':'red'})
    sns.stripplot(data=velocities, x='state', y='velocity', color='black', alpha=0.3, size=1)
    plt.title(f"Eye velocity across states — {sess_name}")
    sns.despine()
    plt.tight_layout()
    plt.show()

    # plot heatmap of eye velocity along state detection and barplots 
    # Decide x-ticks every 100 seconds
    total_frames = len(eye_velocity)
    
    tick_interval_sec = 100
    tick_positions = np.arange(0, total_frames+1, tick_interval_sec * framerate)
    tick_labels = [f"{int(pos/framerate)}" for pos in tick_positions]  # seconds
    
    fig = plt.figure(figsize=(12,4), dpi=300)
    gs = fig.add_gridspec(3, 2, width_ratios=[3,1], height_ratios=[0.15 ,0.03, 0.03], wspace=0.3, hspace=0.05)
    
    # 1. Heatmap
    ax_heat = fig.add_subplot(gs[0,0])
    sns.heatmap(
        eye_velocity_smooth[np.newaxis, :],
        ax=ax_heat,
        cmap='hot',
        cbar=False,
        xticklabels=False,
        yticklabels=True
    )
    # ax_heat.set_xlim(0, len(eye_velocity_smooth))  # align with patch below
    # ax_heat.set_ylabel("eye velocity\n (pixels/s)")
    # ax_heat.set_title("eye motion velocity")

    # norm = plt.Normalize(vmin=np.min(eye_velocity_smooth), vmax=np.max(eye_velocity_smooth))
    # sm = plt.cm.ScalarMappable(cmap='hot', norm=norm)
    # sm.set_array([])
    # cbar = fig.colorbar(sm, cax=ax_cbar)
    # cbar.set_label('Velocity\n (pixels/sec)')
        
    # 2. Behavioral patch
    ax_patch = fig.add_subplot(gs[1,0], sharex=ax_heat)
    for i in range(len(eye_velocity_smooth)):
        if twitch_mot_en[i]:
            ax_patch.axvspan(i, i+1, color='red', alpha=1)
        elif bin_mot_en[i]:
            ax_patch.axvspan(i, i+1, color='orange', alpha=0.5)
        else:
            ax_patch.axvspan(i, i+1, color='blue', alpha=0.5)


    # Remove ticks & labels completely
    ax_patch.set_xlim(0, len(eye_velocity_smooth))
    ax_patch.set_xticks([])
    ax_patch.set_xticklabels([])
    ax_patch.set_ylabel("")  # optional
        
    # 3. Eye velocity trace
    ax_trace = fig.add_subplot(gs[2,0], sharex=ax_heat)
    ax_trace.plot(mot_en_smooth, lw=1)
    ax_trace.set_ylabel("mot_en (au)")
    ax_trace.set_xlabel("Time (s)")
    ax_trace.set_xticks(tick_positions)
    ax_trace.set_xticklabels(tick_labels, fontsize=10)
    
    # 4. Barplot of mean velocity per state
    ax_bar = fig.add_subplot(gs[:,1])
    sns.barplot(data=velocities, x='state', y='velocity', 
                palette={'rest':'blue', 'active':'orange', 'twitch':'red'}, ax=ax_bar)
    ax_bar.set_ylabel("Mean velocity\n (pixels/sec)")
    ax_bar.set_xlabel("")
    ax_bar.set_title("Mean eye velocity per state")
    
    ax_heat.set_aspect('auto')
    ax_patch.set_aspect('auto')
    ax_trace.set_aspect('auto')
        
    sns.despine(fig=fig)
    plt.tight_layout()

    for ax in [ax_heat, ax_patch]:
        for spine in ax.spines.values():
            spine.set_visible(False)
            
    plt.savefig(os.path.join(save_dir, f'{sess_name}_eye_velocity_heatmap_comparisons_per_state.png'))
    plt.show()
    
    
    # Store results
    results.append({
        'ds': sess_name,
        'mouse': mouse_id,
        'age': age,        
        'mot_en': mot_en, 
        'bin_mot_en': bin_mot_en, 
        'eye_displacement_smooth': eye_disp_smooth,
        'eye_velocity_smooth': eye_velocity_smooth,
        'eye_x': eye_x, 
        'eye_y': eye_y,
        'eye_x_velocity': x_velocity,
        'eye_y_velocity': y_velocity,
        'mean_dx': mean_dx,
        'mean_dy': mean_dy, 
        'mean_dx_active': mean_dx_active,
        'mean_dy_active': mean_dy_active,
        'mean_dx_rest': mean_dx_rest,
        'mean_dy_rest': mean_dy_rest, 
        'pearson_r': pear_corr,
        'pearson_active': pear_active_mean,
        'pearson_rest': pear_rest_mean,
        'pearson_twitch': pear_twitch_mean,
        'freq_active': freq_active,
        'freq_rest': freq_rest,
        'peaks_active': peaks_active,
        'peaks_rest': peaks_rest,
        'cross_corr': cross_corr[mask],
        'cross_corr_lags': lags_sec[mask],
        'twitch_max_disp_mean': np.nanmean(max_disp_list),
        'twitch_mean_disp_mean': np.nanmean(mean_disp_list),
        'twitch_fraction_moved': fraction_moved,
        'active_max_disp_mean': np.nanmean(active_max_disp_list),
        'active_mean_disp_mean': np.nanmean(active_mean_disp_list),
        'active_fraction_moved': active_fraction_moved,
        'fraction_eye_followed_twitch_onset': fraction_eye_followed_twitch_onset,
        'fraction_eye_followed_active_onset': fraction_eye_followed_active_onset,
        'fraction_eye_preceded_twitch_onset' : fraction_eye_preceded_twitch_onset,
        'fraction_eye_preceded_active_onset' : fraction_eye_preceded_active_onset,
        'eye_state_velocities': velocities, 
        'mean_vel_twitch':mean_vel_twitch,
        'mean_vel_rest':mean_vel_rest,
        'mean_vel_active': mean_vel_active

                
    })


# Save results as DataFrame
corr_df = pd.DataFrame(results)
print(f"Processed {len(corr_df)} sessions.")
```

```python
from scipy.stats import gaussian_kde
from matplotlib.cm import Blues, Oranges, Grays

# --- Age extraction & color mapping ---
def extract_age_num(label):
    match = re.search(r'\d+', str(label))
    return int(match.group(0)) if match else -1

corr_df['age_num'] = corr_df['age'].apply(extract_age_num)
corr_df = corr_df.sort_values('age_num')
unique_ages = sorted(corr_df['age'].unique(), key=extract_age_num)

min_color = 0.3

age_colors = [Grays(min_color + (1-min_color)*i / max(1, len(unique_ages)-1))
              for i in range(len(unique_ages))] #[Grays(i / max(1, len(unique_ages)-1)) for i in range(len(unique_ages))]
age2color = dict(zip(unique_ages, age_colors))


# --- Function to plot slim half-violins , use that or the above plot_metric_by_age 
def plot_half_violin(df, metric, y_label, title, colors):
    fig, ax = plt.subplots(figsize=(3,3), dpi=300)
    for i, age in enumerate(unique_ages):
        vals = df[df['age']==age][metric].dropna().values
        if len(vals) < 2:
            continue
        mean_val = np.mean(vals)
        sd = np.std(vals)
        kde = gaussian_kde(vals)
        y_grid = np.linspace(mean_val - 2*sd, mean_val + 2*sd, 200)
        density = kde(y_grid)
        density /= density.max()
        width = 0.15 * density
        ax.fill_betweenx(
            y_grid,
            i + 1 - width,  # slim, left-right symmetric
            i + 1 + width,
            facecolor=colors[i],
            alpha=0.3,
            edgecolor='gray',
            linewidth=0.3
        )
        ax.scatter(i+1, mean_val, color='black', s=2, zorder=3)
    ax.axhline(0, linestyle='--', color='lightgrey', lw=2)
    ax.set_xticks(range(1, len(unique_ages)+1))
    ax.set_xticklabels(unique_ages)
    ax.set_ylabel(y_label, fontsize=10)
    ax.set_title(title, y=1.25)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    plt.ylim(-0.2,1)
    plt.savefig(os.path.join(save_dir, f'{title}.png'), dpi=300)
    plt.tight_layout()
    plt.show()

# weighted mixyure od distributions would be more like avg of grey so global avg, here no. of statesdiffer and thsi affects variances and grey is towards blue becs rest dominates

# use to plot one or many variables as afucntion of age
def plot_violin(df, col_list, unique_ages, colors, offsets=None,
                   y_label=None, title="Violin ± SD", save_dir=None):
    """
    Plot violin-like distributions (KDE) for multiple metrics per age.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame containing the columns in `col_list` and 'age'.
    col_list : list of str
        Column names to plot (e.g., ['freq_active', 'freq_rest']).
    unique_ages : list
        List of ages to plot on x-axis.
    colors : list
        Colors for each column/distribution.
    offsets : list of float, optional
        x-offsets to separate distributions (length = len(col_list)).
    y_label : str
        Y-axis label.
    title : str
        Plot title.
    save_dir : str, optional
        Directory to save figure as PNG. If None, figure is not saved.
    """
    fig, ax = plt.subplots(figsize=(4,3), dpi=300)
    
    if offsets is None:
        offsets = np.linspace(-0.15, 0.15, len(col_list))
    
    # Compute global min/max for y-axis
    all_vals = np.concatenate([df[col].dropna().values for col in col_list])
    y_min, y_max = 0, np.max(all_vals)
    
    for i, age in enumerate(unique_ages):
        for col, color, offset in zip(col_list, colors, offsets):
            vals = df.loc[df['age']==age, col].dropna().values
            if len(vals) < 2:
                continue

            mean_val = np.mean(vals)
            sd_val = np.std(vals)

            # KDE for violin shape
            kde = gaussian_kde(vals)
            y_grid = np.linspace(mean_val - 2*sd_val, mean_val + 2*sd_val, 200)
            density = kde(y_grid)
            density /= density.max()  # normalize to 0-1
            width = 0.1 * density  # max half-width

            # Draw filled violin
            ax.fill_betweenx(
                y_grid,
                i + 1 + offset - width,
                i + 1 + offset + width,
                facecolor=color,
                alpha=0.4,
                edgecolor='gray',
                linewidth=0.3
            )

            # Draw mean as dot
            ax.scatter(i+1 + offset, mean_val, color='black', s=10, zorder=3)

    # Axis styling
    ax.axhline(0, linestyle='--', color='lightgrey', lw=1)
    ax.set_xticks(range(1, len(unique_ages)+1))
    ax.set_xticklabels(unique_ages)
    ax.set_ylabel(y_label)
    ax.set_title(title, y=1.1)

    ax.set_ylim(y_min - 0.1*(y_max-y_min), y_max + 0.1*(y_max-y_min))

    sns.despine(ax=ax, left=False, bottom=False)
    plt.tight_layout()

    if save_dir is not None:
        os.makedirs(save_dir, exist_ok=True)
        plt.savefig(os.path.join(save_dir, f"{title}.png"), dpi=300)
    
    plt.show()

# weighted mixture of distributions would be more like avg of grey so global avg, here no. of statesdiffer and thsi affects variances and grey is towards blue becs rest dominates
def plot_violin(df, col_list, unique_ages, colors, offsets=None,
                   y_label=None, title="Violin ± SD", save_dir=None):
    """
    Plot violin-like distributions (KDE) for multiple metrics per age.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame containing the columns in `col_list` and 'age'.
    col_list : list of str
        Column names to plot (e.g., ['freq_active', 'freq_rest']).
    unique_ages : list
        List of ages to plot on x-axis.
    colors : list
        Colors for each column/distribution.
    offsets : list of float, optional
        x-offsets to separate distributions (length = len(col_list)).
    y_label : str
        Y-axis label.
    title : str
        Plot title.
    save_dir : str, optional
        Directory to save figure as PNG. If None, figure is not saved.
    """
    fig, ax = plt.subplots(figsize=(4,3), dpi=300)
    
    if offsets is None:
        offsets = np.linspace(-0.15, 0.15, len(col_list))
    
    # Compute global min/max for y-axis
    all_vals = np.concatenate([df[col].dropna().values for col in col_list])
    y_min, y_max = np.min(all_vals), np.max(all_vals)
    
    for i, age in enumerate(unique_ages):
        for col, color, offset in zip(col_list, colors, offsets):
            vals = df.loc[df['age']==age, col].dropna().values
            if len(vals) < 2:
                continue

            mean_val = np.mean(vals)
            sd_val = np.std(vals)

            # KDE for violin shape
            kde = gaussian_kde(vals)
            y_grid = np.linspace(mean_val - 2*sd_val, mean_val + 2*sd_val, 200)
            density = kde(y_grid)
            density /= density.max()  # normalize to 0-1
            width = 0.1 * density  # max half-width

            # Draw filled violin
            ax.fill_betweenx(
                y_grid,
                i + 1 + offset - width,
                i + 1 + offset + width,
                facecolor=color,
                alpha=0.4,
                edgecolor='gray',
                linewidth=0.3
            )

            # Draw mean as dot
            ax.scatter(i+1 + offset, mean_val, color='black', s=10, zorder=3)

    # Axis styling
    ax.axhline(0, linestyle='--', color='lightgrey', lw=1)
    ax.set_xticks(range(1, len(unique_ages)+1))
    ax.set_xticklabels(unique_ages)
    ax.set_ylabel(y_label)
    ax.set_title(title, y=1.1)

    ax.set_ylim(y_min - 0.1*(y_max-y_min), y_max + 0.1*(y_max-y_min))

    sns.despine(ax=ax, left=False, bottom=False)
    plt.tight_layout()

    if save_dir is not None:
        os.makedirs(save_dir, exist_ok=True)
        plt.savefig(os.path.join(save_dir, f"{title}.png"), dpi=300)
    
    plt.show()
```

```python
# Compute peak cross-corr lag per session
corr_df['peak_corr_lag'] = corr_df.apply(
    lambda row: row['cross_corr_lags'][np.argmax(row['cross_corr'])] 
    if row['cross_corr'] is not None and len(row['cross_corr']) > 0 else np.nan, 
    axis=1
)

# Plot mean ± SD per age

fig, ax = plt.subplots(figsize=(4, 3))
max_val = 0

for i, age in enumerate(unique_ages):
    vals = corr_df.loc[corr_df['age'] == age, 'peak_corr_lag'].dropna().values
    if len(vals) > 0:
        mean_val = np.mean(vals)
        sd_val = np.std(vals)
        max_val = max(max_val, mean_val + sd_val)
        
        # vertical line ± SD
        ax.plot([i+1]*2, [mean_val - sd_val, mean_val + sd_val], color='grey', lw=4, alpha=0.6)
        # mean dot
        ax.scatter(i+1, mean_val, color=colors[i], s=20, zorder=3)

ax.axhline(0, linestyle='--', color='gray', lw=1)
ax.set_xticks(range(1, len(unique_ages)+1))
ax.set_xticklabels(unique_ages)
ax.set_yticks(np.linspace(0, max_val, num=4))  # 0, mid, max
ax.set_ylabel('Lag at peak cross-corr (s)')
ax.set_title('Peak cross-corr lag per age', y=1.1)
sns.despine(ax=ax)
plt.tight_layout()
plt.show()
```

```python
# --- Peak lag vs age ---
peak_lags = []
ages_for_plot = []

for age in unique_ages:
    sub = corr_df.loc[corr_df['age'] == age]
    if len(sub) == 0:
        continue

    # Compute average cross-correlation
    avg_cc = np.nanmean(np.stack(sub['cross_corr'].values), axis=0)
    lags_sec = sub['cross_corr_lags'].values[0]  # same lags for all sessions

    # Find lag of max absolute correlation
    peak_idx = np.nanargmax(np.abs(avg_cc))
    peak_lag = lags_sec[peak_idx]

    # Store for plotting
    peak_lags.append(peak_lag)
    ages_for_plot.append(age)

# Plot
fig, ax = plt.subplots(figsize=(5,3.5))
ax.scatter(ages_for_plot, peak_lags, s=80, color=[age2color[a] for a in ages_for_plot], edgecolor='k')
ax.plot(ages_for_plot, peak_lags, linestyle='--', color='gray', alpha=0.5)  # optional connecting line

ax.set_xlabel('Age')
ax.set_ylabel('Lag of max cross-corr (s)')
ax.set_title('Peak lag of eye displacement vs motion energy per age')
sns.despine()
plt.tight_layout()
fig_path = os.path.join(save_dir, 'eye_mot_peak_lag_vs_age.png')
plt.savefig(fig_path, dpi=300)
plt.show()
```

```python
from scipy.stats import gaussian_kde
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- Shared setup ---
unique_ages = sorted(corr_df['age'].unique(), key=extract_age_num)
age_colors = [Blues(i / max(1, len(unique_ages)-1)) for i in range(len(unique_ages))]
age2color = dict(zip(unique_ages, age_colors))

def safe_kde(vals):
    """Return KDE (y_grid, density) safely; None if not computable."""
    vals = np.asarray(vals)
    if len(vals) < 2 or np.allclose(vals, vals[0]):  # Too few or identical
        return None, None
    try:
        kde = gaussian_kde(vals)
        y_grid = np.linspace(min(vals), max(vals), 200)
        density = kde(y_grid)
        density /= density.max()
        return y_grid, density
    except Exception:
        return None, None

# --- Fraction of twitches with eye movement ---
fig, ax = plt.subplots(figsize=(3, 2), dpi=300)
for i, age in enumerate(unique_ages):
    vals = corr_df.loc[corr_df['age'] == age, 'active_fraction_moved'].dropna().values
    if len(vals) == 0:
        continue
    mean_val = np.mean(vals)
    y_grid, density = safe_kde(vals)
    if y_grid is not None:
        width = 0.2 * density
        ax.fill_betweenx(y_grid, i + 1 - width / 2, i + 1 + width / 2,
                         facecolor=age2color[age], alpha=0.5,
                         edgecolor='grey', linewidth=0.4)
    ax.scatter(i + 1, mean_val, color='black', s=8, zorder=3)

ax.set_xticks(range(1, len(unique_ages) + 1))
ax.set_xticklabels(unique_ages)
ax.set_ylabel('% eye mot after awake onset', fontsize=10)
sns.despine()
plt.tight_layout()
plt.ylim(0, 1)
plt.show()

# --- Mean max displacement per age ---
fig, ax = plt.subplots(figsize=(3, 2), dpi=300)
for i, age in enumerate(unique_ages):
    vals = corr_df.loc[corr_df['age'] == age, 'active_max_disp_mean'].dropna().values
    if len(vals) == 0:
        continue
    mean_val = np.mean(vals)
    y_grid, density = safe_kde(vals)
    if y_grid is not None:
        width = 0.2 * density
        ax.fill_betweenx(y_grid, i + 1 - width / 2, i + 1 + width / 2,
                         facecolor=age2color[age], alpha=0.5,
                         edgecolor='grey', linewidth=0.4)
    ax.scatter(i + 1, mean_val, color='black', s=8, zorder=3)

ax.set_xticks(range(1, len(unique_ages) + 1))
ax.set_xticklabels(unique_ages)
ax.set_ylabel('max disp. after awake onset', fontsize=10)
sns.despine()
plt.tight_layout()
plt.ylim(0, 1)
plt.show()
```

```python
# 
col_map = {'rest': 'mean_vel_rest', 'active': 'mean_vel_active', 'twitch': 'mean_vel_twitch'}

# Melt dataframe for seaborn
mean_velocities = full_df[['mean_vel_twitch', 'mean_vel_rest', 'mean_vel_active']].copy()
mean_velocities_melted = mean_velocities.melt(var_name='state', value_name='velocity')
mean_velocities_melted['state'] = mean_velocities_melted['state'].replace({
    'mean_vel_rest': 'rest',
    'mean_vel_active': 'active',
    'mean_vel_twitch': 'twitch'
})

# Barplot
plt.figure(figsize=(4,4), dpi=300)
ax = sns.barplot(
    data=mean_velocities_melted, x='state', y='velocity',
    palette={'rest':'blue', 'active':'orange', 'twitch':'red'}, ci='sd'
)
plt.ylabel('Eye velocity\n(pixels/s)')
plt.title('Mean eye velocity per state', y=1.1)

# Paired t-tests and significance
states = ['rest', 'active', 'twitch']
comparisons = [('rest','active'), ('rest','twitch'), ('active','twitch')]
y_max = mean_velocities_melted['velocity'].max() + 2
h = 0.5

for i, (s1, s2) in enumerate(comparisons):
    data1 = mean_velocities[col_map[s1]]
    data2 = mean_velocities[col_map[s2]]
    
    # Remove any NaNs just in case
    mask = (~data1.isna()) & (~data2.isna())
    t_stat, p_val = ttest_rel(data1[mask], data2[mask])
    
    # Determine significance symbol
    if p_val < 0.0005: sig = '***'
    elif p_val < 0.005: sig = '**'
    elif p_val < 0.05: sig = '*'
    else: sig = 'n.s.'
    
    print(f"{s1} vs {s2}: t={t_stat:.3f}, p={p_val:.5f}, sig={sig}")
    
    # Draw significance lines
    x1, x2 = states.index(s1), states.index(s2)
    y = y_max + i*3
    ax.plot([x1, x1, x2, x2], [y, y+h, y+h, y], c='black')
    ax.text((x1+x2)/2, y+h, sig, ha='center', va='bottom', color='black')

sns.despine()
plt.tight_layout()
plt.savefig(os.path.join(save_dir, 'eye_velocity_per_state.png'))
plt.show()
```
